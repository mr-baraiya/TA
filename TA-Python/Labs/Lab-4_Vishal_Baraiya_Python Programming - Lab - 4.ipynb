```python
n = int(input("Enter the number : "))
for i in range(1, n+1):
    print(i)
```

## 3. WAP to print odd numbers between 1 to n.

```python
n = int(input("Enter the number : "))
for i in range(1, n+1):
    if i % 2 != 0:
        print(i)
```

## 4. WAP to print numbers between two given numbers which are divisible by 2 but not divisible by 3.

```python
start = int(input("Enter the start number : "))
end = int(input("Enter the end number : "))
for i in range(start, end+1):
    if i % 2 == 0 and i % 3 != 0:
        print(i)
```

## 5. WAP to print sum of 1 to n numbers.

```python
n = int(input("Enter the number : "))
s = 0
for i in range(1, n+1):
    s += i
print('Sum =', s)
```

## 8. WAP to print sum of series 1 – 2 + 3 – 4 + 5 – 6 + ... n.

```python
n = int(input("Enter the number : "))
s = 0
sign = 1
for i in range(1, n+1):
    s += i * sign
    sign *= -1
print('Series sum =', s)
```

## 9. WAP to find factorial of the given number.

```python
n = int(input("Enter the number : "))
fact = 1
for i in range(1, n+1):
    fact *= i
print('Factorial =', fact)
```

## 10. WAP to find whether the given number is prime or not.

```python
n = int(input("Enter the number : "))
if n < 2:
    print('Not Prime')
else:
    for i in range(2, n//2+1):
        if n % i == 0:
            print('Not Prime')
            break
    else:
        print('Prime')
```

## 12. WAP to print sum of digits of a given number.

```python
n = int(input("Enter the number : "))
s = 0
while n > 0:
    s += n % 10
    n //= 10
print('Sum of digits =', s)
```

## 14. WAP to check whether the given number is palindrome or not.

```python
num = int(input("Enter a number: "))
temp = num
rev = 0

while temp > 0:
    digit = temp % 10
    rev = rev * 10 + digit
    temp //= 10

if rev == num:
    print("Palindrome")
else:
    print("Not Palindrome")

# n = int(input("Enter the number : "))
# if str(n) == str(n)[::-1]:
#     print('Palindrome')
# else:
#     print('Not Palindrome')
```
