### 02) WAP to reverse the words in the given string.

```python
s = input("Enter the String : ")
words = s.split()
rev = ' '.join(words[::-1])
print(rev)
```

### 04) WAP to find length of string without using len function.

```python
s = input("Enter the String : ")
count = 0
for _ in s:
    count += 1
print('Length =', count)
```

### 05) WAP to print even length word in string.

```python
s = input("Enter the String : ")
for word in s.split():
    if len(word) % 2 == 0:
        print(word)
```

### 06) WAP to count numbers of vowels in given string.

```python
s = input("Enter the String : ")
vowels = 'aeiouAEIOU'
count = 0
for ch in s:
    if ch in vowels:
        count += 1
print('Vowel count =', count)
```

### 07) WAP to capitalize the first and last character of each word in a string.

```python
s = input("Enter the String : ")
res = []
for w in s.split():
    if len(w) > 1:
        w = w[0].upper() + w[1:-1] + w[-1].upper()
    else:
        w = w.upper()
    res.append(w)
print(' '.join(res))
```

### 09) Check if the password and confirm password is same or not. 
### In case of only case's mistake, show the error message.

```python
password = input("Enter password: ")
confirm_password = input("Confirm password: ")

if password == confirm_password:
    print("Password matched successfully!")
elif password.lower() == confirm_password.lower():
    print("Passwords do not match due to case sensitivity!")
else:
    print("Passwords do not match!")
```

### 10) : Display credit card number. 
### card no. : 1234 5678 9012 3456
### display as  : **** **** **** 3456

```python
card_no = "1234 5678 9012 3456"

cleaned_card = card_no.replace(" ", "")

masked = "*" * (len(cleaned_card) - 4) + cleaned_card[-4:]

formatted = " ".join([masked[i:i+4] for i in range(0, len(masked), 4)])

print("Original card no. :", card_no)
print("Display as        :", formatted)
```

### 11) : Checking if the two strings are Anagram or not.
### s1 = decimal and s2 = medical are Anagram

```python
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

s1 = s1.replace(" ", "").lower()
s2 = s2.replace(" ", "").lower()

# Checking anagram
if sorted(s1) == sorted(s2):
    print("The strings are Anagrams")
else:
    print("The strings are NOT Anagrams")
```

### 12) : Rearrange the given string. First lowercase then uppercase alphabets.
### input : EHlsarwiwhtwMV
### output : lsarwiwhtwEHMV

```python
s = input("Enter string: ")

lowercase = ""
uppercase = ""

for ch in s:
    if ch.islower():
        lowercase += ch
    elif ch.isupper():
        uppercase += ch

result = lowercase + uppercase
print("Rearranged string:", result)
```
