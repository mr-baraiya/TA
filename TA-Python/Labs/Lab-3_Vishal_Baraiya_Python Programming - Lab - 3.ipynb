## 8. WAP to read marks of five subjects and print class (Fail/Pass/Second/First/Distinction).

```python
marks = []
for i in range(5):
    marks.append(float(input(f"Enter mark {i+1}: ")))
ans = 0
for i in marks:
    ans += i
p = ans / 5

if p < 35:
    print('Fail')
elif p <= 45:
    print('Pass Class')
elif p <= 60:
    print('Second Class')
elif p <= 70:
    print('First Class')
else:
    print('Distinction')
```

## 9. WAP to check whether a triangle is isosceles, equilateral, scalene or right-angled.

```python
a = float(input("Enter side a: "))
b = float(input("Enter side b: "))
c = float(input("Enter side c: "))

if a == b == c:
    print('Equilateral')
elif a == b or b == c or a == c:
    print('Isosceles')
elif a*a + b*b == c*c or a*a + c*c == b*b or b*b + c*c == a*a:
    print('Right Angled')
else:
    print('Scalene')
```

## 10. WAP to find the second largest number among three user input numbers.

```python
a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

if a > b and a > c:
    second = b if b > c else c
elif b > a and b > c:
    second = a if a > c else c
else:
    second = a if a > b else b

print("Second largest =", second)
```

## 11. WAP to calculate electricity bill based on given unit criteria.
### i. First 1 to 50 units – Rs. 2.60/unit  
### ii. Next 50 to 100 units – Rs. 3.25/unit 
### iii. Next 100 to 200 units – Rs. 5.26/unita 
### iv. bove 200 units – Rs. 8.45/unit 

```python
units = float(input("Enter units: "))
bill = 0

if units <= 50:
    bill = units * 2.60
elif units <= 100:
    bill = 50 * 2.60 + (units - 50) * 3.25
elif units <= 200:
    bill = 50 * 2.60 + 50 * 3.25 + (units - 100) * 5.26
else:
    bill = 50 * 2.60 + 50 * 3.25 + 100 * 5.26 + (units - 200) * 8.45

print('Bill =', bill)
```
