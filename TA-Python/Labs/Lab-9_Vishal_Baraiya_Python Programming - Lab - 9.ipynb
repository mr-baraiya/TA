# 03) Write a function that returns 1 if the given number is Prime or 0 otherwise.

```python
def isPrime(n):
    count = 0
    for i in range(1,n+1):
        if(n % i == 0):
            count += 1
        else:
            pass
    if count == 2 :
        return 1
    else:
        return 0

n = int(input("Enter the Number : "))
if isPrime(n):
    print(f"{n} is a Prime Number.")
else:
    print(f"{n} is Not a Prime Number.")
```

### 04) Write a function that returns the list of Prime numbers between given two numbers.

```python
def isPrime(n):
    count = 0
    for i in range(1,n+1):
        if(n % i == 0):
            count += 1
        else:
            pass
    if count == 2 :
        return True
    else:
        return False

def findPrimeInRange (start,end):
    l = []
    for i in range(start,end):
        if isPrime(i):
            l.append(i)
        else:
            pass
    return l
start = int(input("Enter the Start : "))
end = int(input("Enter the end : "))
l = findPrimeInRange(start,end)
print(l)
```

### 05) Write a function that returns True if the given string is Palindrome or False otherwise.

```python
def isPalindrome(s):
    s = s.lower()
    return s == s[::-1]

s = input("Enter the String : ")
if isPalindrome(s):
    print(f"{s} is a Palindrome string.")
else:
    print(f"{s} is Not a Palindrome string.")
```

### 11) Write a function to invert a given Dictionary.
#### hint: keys to values & values to keys
#### Before : {'a': 10, 'b':20, 'c':30, 'd':40}
#### After : {10:'a', 20:'b', 30:'c', 40:'d'}

```python
def invertDictionary(d):
    di = {}
    for i in d.keys():
        di[d[i]] = i
    return di

d = {'a': 10, 'b':20, 'c':30, 'd':40}
print("Before : ",d)
d = invertDictionary(d)
print("After : ",d)
```

### 12) Write a function to check whether the given string is Pangram or not.
#### hint: Pangram is a string containing all the characters a-z atlest once.
#### "the quick brown fox jumps over the lazy dog" is a Pangram string.

```python
def isPanagram(s):
    s=s.replace(" ","")
    s=s.lower()
    s=list(set(s))
    s.sort()
    s="".join(s)
    alphabets = "abcdefghijklmnopqrstuvwxyz"
    return s == alphabets
s = "the quick brown fox jumps over the lazy dog"
print(f"{isPanagram(s)}")
```

### 13) Write a function that returns the number of uppercase and lowercase letters in the given string. 
#### example : Input : s1 = AbcDEfgh ,Ouptput : no_upper = 3, no_lower = 5

```python
def count_lower(s : str):
    countLower = 0
    for i in s:
        if i.islower():
            countLower += 1
    return countLower
def count_upper(s : str):
    countUpper = 0
    for i in s:
        if i.isupper():
            countUpper += 1
    return countUpper

s1 = "AbcDEfgh"
print(f"Count Of Lower : {count_lower(s1)}")
print(f"Count Of Upper : {count_upper(s1)}")
```

### 17) Write udfs to call the functions with following types of arguments:
1. Positional Arguments
2. Keyword Arguments
3. Default Arguments
4. Variable Legngth Positional(*args) & variable length Keyword Arguments (**kwargs)
5. Keyword-Only & Positional Only Arguments

```python
def positional(a, b):
    print("Positional:", a, b)

positional(10, 20)

def keyword(a, b):
    print("Keyword:", a, b)

keyword(b=20, a=10)

def default(a, b=50):
    print("Default:", a, b)

default(10)
default(10, 100)

def var_positional(*args):
    print("Variable Length Positional (*args):", args)

var_positional(1, 2, 3, 4)

def var_keyword(**kwargs):
    print("Variable Length Keyword (**kwargs):", kwargs)

var_keyword(name="Vishal", age=20, city="Delhi")

def pos_only(a, b, /):
    # Parameters before '/' must be positional only
    print("Positional-Only:", a, b)

pos_only(5, 10)
# pos_only(a=5, b=10)  # Error

def kw_only(*, a, b):
    # Parameters after '*' must be passed using keyword
    print("Keyword-Only:", a, b)

kw_only(a=5, b=10)
# kw_only(5, 10)      # Error
```

```python
def example(a, b, /, c, d, *, e, f):
    """
    a, b → Positional-only (must be passed without using names)
    c, d → Can be positional or keyword
    e, f → Keyword-only (must be passed using names)
    """
    print("Positional-only (a, b):", a, b)
    print("Pos/Key arguments (c, d):", c, d)
    print("Keyword-only (e, f):", e, f)

example(10, 20, 30, d=40, e=50, f=60)
```
