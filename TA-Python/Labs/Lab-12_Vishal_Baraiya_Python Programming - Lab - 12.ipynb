## 1. WAP to store student names in a list. Use list methods to add names, insert a new name at a specific position, remove a name, sort the list, reverse it, and display the final result.

```python
students = []

students.append("Rahul")
students.append("Sneha")
students.append("Vishal")
students.append("Priya")
students.append("Aman")

print("Initial List:", students)

students.insert(2, "Karan")
print("After Inserting 'Karan' at position 2:", students)

students.remove("Aman")
print("After Removing 'Aman':", students)

students.sort()
print("After Sorting:", students)

students.reverse()
print("After Reversing:", students)

print("\nFinal Student List:", students)
```

## 2. WAP to merge two lists and sort the combined list and then find how many times a particular value appears.

```python
list1 = list(map(int, input("Enter elements of list1 separated by space: ").split()))
list2 = list(map(int, input("Enter elements of list2 separated by space: ").split()))

merged_list = list1 + list2

merged_list.sort()

value = int(input("Enter the value to count: "))

count = merged_list.count(value)

print("\nMerged List:", merged_list)
print(f"The value '{value}' appears {count} times in the merged sorted list.")
```

## 3. WAP to insert a given value at every even index of a list and display the final list.

```python
my_list = list(map(int, input("Enter elements of the list separated by space: ").split()))

value = int(input("Enter the value to insert at every even index: "))

i = 0
while i < len(my_list):
    my_list.insert(i, value)
    i += 2

print("Final List:", my_list)
```

## 4. Unique Word Analyzer — Read article.txt, extract words, make set for unique words, create frequency dictionary, store result in wordcount.txt.

```python
import string

with open('article.txt', 'r') as file:
    text = file.read()

text = text.lower()
text = text.translate(str.maketrans("", "", string.punctuation))

words = text.split()

unique_words = set(words)

frequency = {}
for word in words:
    frequency[word] = frequency.get(word, 0) + 1

with open('wordcount.txt', 'w') as output:
    output.write("Unique Words:\n")
    output.write(str(unique_words) + "\n\n")

    output.write("Word Frequency:\n")
    for word, count in frequency.items():
        output.write(f"{word}: {count}\n")

print("Analysis completed! Results stored in 'wordcount.txt'.")
```

## 5. Customer Purchase Summary — Read purchases.txt, remove duplicate amounts, compute average, create dictionary using CustomerID, store result in summary.txt using exception handling.

```python
try:
    with open('purchases.txt', 'r') as file:
        lines = file.readlines()

    customer_data = {}

    for line in lines:
        parts = line.strip().split()
        if len(parts) == 2:
            customer_id, amount = parts[0], float(parts[1])
            if customer_id not in customer_data:
                customer_data[customer_id] = []
            customer_data[customer_id].append(amount)

    final_summary = {}
    for cust_id, amounts in customer_data.items():
        unique_amounts = list(set(amounts))
        average = sum(unique_amounts) / len(unique_amounts)
        final_summary[cust_id] = {"Unique Purchases": unique_amounts, "Average": average}

    with open('summary.txt', 'w') as output:
        output.write("Customer Purchase Summary:\n")
        for cust_id, data in final_summary.items():
            output.write(f"CustomerID: {cust_id}, Purchases (Unique): {data['Unique Purchases']}, Average: {data['Average']}\n")

    print("Summary generated successfully in 'summary.txt'.")

except FileNotFoundError:
    print("Error: purchases.txt file not found.")
except ValueError:
    print("Error: Invalid data format in purchases.txt.")
except Exception as e:
    print("An unexpected error occurred:", e)
```
